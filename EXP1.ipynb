```python
# FIND-S Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Cool', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes']
]

hypothesis = ['0', '0', '0', '0', '0', '0']

for sample in data:
    if sample[-1] == 'Yes':
        if hypothesis == ['0', '0', '0', '0', '0', '0']:
            hypothesis = sample[:-1].copy()
        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != sample[i]:
                    hypothesis[i] = '?'

print("Training Data:")
for row in data:
    print(row)

print("\nFinal Hypothesis:")
print(hypothesis)
```
