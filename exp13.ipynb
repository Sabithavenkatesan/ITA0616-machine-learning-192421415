```python

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Create dataset
data = {
    'Year': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022],
        'Kilometers': [80000, 70000, 60000, 50000, 40000, 30000, 20000, 10000],
            'Price': [300000, 350000, 400000, 450000, 500000, 550000, 600000, 650000]
            }

            df = pd.DataFrame(data)

            # Input and output
            X = df[['Year', 'Kilometers']]
            y = df['Price']

            # Create model
            model = LinearRegression()

            # Train model
            model.fit(X, y)

            # Predict
            y_pred = model.predict(X)

            # Display output
            print("Actual Prices:", y.values)
            print("Predicted Prices:", y_pred)
            print("Mean Squared Error:", mean_squared_error(y, y_pred))

            # Predict new car price
            new_car = [[2023, 15000]]
            prediction = model.predict(new_car)

            print("\nPredicted Car Price:", prediction[0])
```
