```python
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

S = ['0', '0', '0', '0', '0', '0']
G = [['?', '?', '?', '?', '?', '?']]

print("Training Data:")
for row in data:
    print(row)

for row in data:
    x = row[:6]
    label = row[6]

    if label == 'Yes':
        for i in range(6):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

        # Filter G to keep only hypotheses consistent with the positive instance
        G = [g for g in G if all(g[i] == '?' or g[i] == S[i] for i in range(6))]
    else:
        new_G = []
        for g in G:
            for i in range(6):
                if S[i] != '0' and S[i] != '?' and S[i] != x[i]:
                    h = g.copy()
                    h[i] = S[i]
                    if h not in new_G:
                        new_G.append(h)
        G = new_G

print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for h in G:
    print(h)
```
