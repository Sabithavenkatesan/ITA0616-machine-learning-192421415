```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture

# Generate dataset
X, y = make_blobs(n_samples=150, centers=3, random_state=42)

# Create Gaussian Mixture Model
model = GaussianMixture(n_components=3, random_state=42)

# Train model using EM algorithm
model.fit(X)

# Predict cluster labels
labels = model.predict(X)

# Display output
print("Cluster Labels:")
print(labels)

print("\nCluster Centers:")
print(model.means_)

print("\nNumber of Clusters:", model.n_components)
print("Converged:", model.converged_)
print("Iterations:", model.n_iter_)

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(model.means_[:, 0], model.means_[:, 1],
            marker='X', s=200)
            plt.title("Expectation Maximization (EM) Clustering")
            plt.xlabel("Feature 1")
            plt.ylabel("Feature 2")
            plt.show()
```
